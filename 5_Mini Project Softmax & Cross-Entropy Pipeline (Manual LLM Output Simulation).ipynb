```python
import numpy as np
```

```python
# Step 1: Define vocabulary (like a mini token set)
vocab = ["apple", "banana", "cat", "dog", "elephant"]

# Step 2: Simulate model logits (raw scores output by a fake model)
logits = np.array([0.2, 3.5, 0.2, -0.5, 0.7])  # Notice "banana" has the highest score now

# Step 3: Apply softmax to convert logits to probabilities
def softmax(z):
    z_stable = z - np.max(z)  # for numerical stability
    exp_z = np.exp(z_stable)
    return exp_z / np.sum(exp_z)

probs = softmax(logits)

# Step 4: Prediction
predicted_index = np.argmax(probs)
predicted_word = vocab[predicted_index]

# Step 5: Define the true label (correct answer is now "banana")
true_index = vocab.index("banana")
true_one_hot = np.zeros(len(vocab))
true_one_hot[true_index] = 1

# Step 6: Compute cross-entropy loss
def cross_entropy(y_true, y_pred):
    return -np.sum(y_true * np.log(y_pred + 1e-10))  # small epsilon to avoid log(0)

loss = cross_entropy(true_one_hot, probs)

# Print everything
print("Vocabulary:", vocab)
print("Logits:    ", logits)
print("Probabilities (softmax):", np.round(probs, 3))
print("Prediction Index:", predicted_index)
print("Prediction:", predicted_word)
print("true_index:", true_index)
print("true_one_hot:", true_one_hot)
print("True label:", vocab[true_index])
print("Cross-Entropy Loss:", round(loss, 4))
```
