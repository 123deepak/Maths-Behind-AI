```python
import math
import matplotlib.pyplot as plt
```

```python
x_vals = [i for i in range(1, 101)]
log10_vals = [math.log10(x) for x in x_vals]
ln_vals = [math.log(x) for x in x_vals]

plt.figure(figsize=(10,6))
plt.plot(x_vals, log10_vals, label='log10(x)', marker='o')
plt.plot(x_vals, ln_vals, label='ln(x)', marker='+')
plt.xlabel('x')
plt.ylabel('log value')
plt.title('log10(x) vs ln(x)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# we preper log base e (natural log) instead of just log base 10
# as it aligns more with exponential data nature and simplifies the calculation.
```
