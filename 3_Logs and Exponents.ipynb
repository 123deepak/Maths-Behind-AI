```python
import numpy as np
import matplotlib.pyplot as plt

# Choose a number
x = 2.5

# Apply exponential, then log
exp_then_log = np.log(np.exp(x))

# Apply log, then exponential
log_then_exp = np.exp(np.log(x))

print(f"Original x: {x}")
print(f"log(exp(x)) = {exp_then_log}")  # Should be close to x
print(f"exp(log(x)) = {log_then_exp}")  # Should be close to x
```

__1. Why do we use the exponential function (e^x) in softmax?__

Using the exponential function serves a few core purposes:

🔹 a) Amplifies differences between values

Exponential grows faster than linear, so:

Small differences in inputs become larger differences after applying exp(x).

This makes the largest input (logit) dominate more clearly in the final probabilities.
So even a small difference in input results in a much more decisive probability gap.

🔹 b) Keeps all outputs positive
e^x is always > 0, which is important because probabilities must be non-negative.

__2. Why divide by the sum of exponentials?__

Because we're trying to turn arbitrary scores into a probability distribution:



iterating through all values using j variable for each i 


🔹 a) Normalization
This ensures that all softmax outputs:

Are between 0 and 1.

Sum up to 1, which is a requirement for valid probabilities.

🔹 b) Probabilistic interpretation
Now each softmax(x_i) can be interpreted as:

"The probability that class i is the correct one, given the current model outputs."

🧠 Intuition Recap
Think of softmax as a "soft" winner-takes-all function:

The largest value gets most of the probability.

Others get less, but still non-zero — allowing for gradient updates and learning.

```python
# Step 1: Generate 50 values between 0 and 5
x = np.linspace(0, 5, 50)

# Step 2: Compute exp(x)
exp_vals = np.exp(x)

# Step 3: Compute log(exp(x)) => should return original x
log_of_exp = np.log(exp_vals)

# Step 4: Plot both
plt.figure(figsize=(10, 6))
plt.plot(x, exp_vals, label='exp(x)', color='blue')
plt.plot(x, log_of_exp, label='log(exp(x))', color='orange')

# Styling
plt.title('exp(x) and log(exp(x))')
plt.xlabel('x')
plt.ylabel('Value')
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```
