```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_vectors(vectors, colors):
    fig, ax = plt.subplots()
    ax.axhline(0, color='gray', linewidth=0.5)
    ax.axvline(0, color='gray', linewidth=0.5)
    
    for i in range(len(vectors)):
        ax.quiver(0, 0, vectors[i][0], vectors[i][1], 
                  angles='xy', scale_units='xy', scale=1, 
                  color=colors[i], label=f'vector {i+1}')
    
    ax.set_xlim(-2, 4)
    ax.set_ylim(-2, 4)
    ax.set_aspect('equal')
    ax.grid()
    ax.legend()
    plt.title('Vector Visualization')
    plt.show()

# Define 2D vectors for clarity
v1 = np.array([2, 2])
v2 = np.array([2, 1.75]) 
v3 = np.array([-2, -1])    
v4 = np.array([0, 1])
v5 = np.array([-2, -2])
v6 = np.array([2, -2])


# Plot
plot_vectors([v1, v2, v3, v4, v5, v6], ['blue', 'green', 'red', 'orange', 'grey', 'pink'])

# Compute dot products
print("v1 · v2 =", np.dot(v1, v2))
print("v1 · v3 =", np.dot(v1, v3))
print("v1 · v4 =", np.dot(v1, v4))
print("v1 · v5 =", np.dot(v1, v5))
print("v1 · v5 =", np.dot(v1, v6))
```

Note:
 if we know theta between two vectors, then we use ||a||.||b||.cos(theta) but if theta is unknown, then we multiply two vectors together to get dot product.  cosine similarity is the computed value of 
cos(𝜃) between two vectors.

💡 Interpretation:
Same direction → dot product is large and positive

Opposite direction → large negative

Perpendicular → near zero

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("Cosine similarity (v1, v2):", cosine_similarity(v1, v2))
print("Cosine similarity (v1, v3):", cosine_similarity(v1, v3))
print("Cosine similarity (v1, v4):", cosine_similarity(v1, v4))
print("Cosine similarity (v1, v5):", cosine_similarity(v1, v5))
print("Cosine similarity (v1, v6):", cosine_similarity(v1, v6))
```
