```python
import matplotlib.pyplot as plt
import numpy as np

# Angle in degrees
theta_deg = np.linspace(0, 45, 500)
theta_rad = np.radians(theta_deg)

# Compute slope = tan(theta)
slopes = np.tan(theta_rad)

# Special points
special_angles = [15, 30, 45]
special_slopes = np.tan(np.radians(special_angles))

plt.figure(figsize=(8, 4))
plt.plot(theta_deg, slopes, color='blue', label='tan(θ)')

# Add markers
for angle, slope in zip(special_angles, special_slopes):
    plt.plot(angle, slope, 'ro')
    plt.text(angle + 0.5, slope, f"{angle}°\n≈ {slope:.2f}", color='red')

plt.title("Slope vs Angle (θ) — m = tan(θ)")
plt.xlabel("Angle θ (degrees)")
plt.ylabel("Slope (m = tan(θ))")
plt.grid(True)
plt.axhline(0, color='gray', lw=0.5)
plt.axvline(0, color='gray', lw=0.5)
plt.ylim(0, 1)
plt.legend()
plt.show()
```
