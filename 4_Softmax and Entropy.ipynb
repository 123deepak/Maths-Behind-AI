A logit is the output of the final linear transformation (e.g., Wx + b) in a neural network, before any activation.

These values are not probabilities — they can be positive, negative, or zero.

After applying softmax to the logits, you get probabilities.

The term originates from logistic regression, where the "logit function" is:


But in deep learning, the term has been adapted to mean:

"The inputs to a softmax layer (before softmax is applied)."

"Raw scores" are often called logits because they are the direct output of a model before any probability transformation.

It's standard jargon in machine learning frameworks like TensorFlow and PyTorch.

## Softmax 

```python
import numpy as np
import matplotlib.pyplot as plt

logits = np.array([4.2, -1.3, 0.7, 2.1])
softmax = np.exp(logits) / np.sum(np.exp(logits))

print("Softmax:", np.round(softmax, 3))
print("Sum of softmax:", np.sum(softmax))
```

```python
x = np.linspace(-5, 5, 10)
softmax_vals = np.exp(x) / np.sum(np.exp(x))

plt.plot(x, softmax_vals)
plt.title("Softmax Curve (1D Slice)")
plt.grid(True)
plt.show()
print(softmax_vals)
```

```python
import numpy as np

def stable_softmax(z):
    # Step 1: Subtract max for numerical stability
    z_stable = z - np.max(z)
    
    # Step 2: Compute exponentials
    exp_z = np.exp(z_stable)
    
    # Step 3: Normalize to get probabilities
    softmax = exp_z / np.sum(exp_z)
    
    return softmax

# Example input
z = np.array([4.2, 1.3, 0.8, -1.0, 0.1])

# Compute softmax
softmax_output = stable_softmax(z)
print("Softmax output:", np.round(softmax_output, 3))
print("Sum of probabilities:", np.sum(softmax_output))  # should be 1.0
```

```python
y_true = np.array([0, 1, 0, 0])
loss = -np.sum(y_true * np.log(softmax))
print("Cross-Entropy Loss:", loss)
```
